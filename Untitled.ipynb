```python
import pandas as pd

cols = ['sendTime', 'sender', 'class',
        'posx', 'posy',
        'spdx', 'spdy',
        'aclx', 'acly',
        'hedx', 'hedy']

dtypes = {
    'sendTime': 'float32', 'sender': 'int32', 'class': 'int8',
    'posx': 'float32', 'posy': 'float32',
    'spdx': 'float32', 'spdy': 'float32',
    'aclx': 'float32', 'acly': 'float32',
    'hedx': 'float32', 'hedy': 'float32',
}

df = pd.read_csv(
    r"C:\projects\v2x-detection\data\extracted\mixalldata_clean.csv",
    usecols=cols, dtype=dtypes
)

print("Rows:", len(df))
print("Memory used:", round(df.memory_usage(deep=True).sum() / 1024**2, 1), "MB")
print()
df.head()
```

```python
import numpy as np

# sendTime values are ~72000, and float32 only resolves ~0.008s at that
# magnitude. Time gaps need better precision than that.
df['sendTime'] = df['sendTime'].astype('float64')

# CRITICAL: sort by sender, then time. Every feature below depends on this.
df = df.sort_values(['sender', 'sendTime']).reset_index(drop=True)

print("Sorted.")
print("Unique vehicles:", df['sender'].nunique())
print("Messages per vehicle (avg):", round(len(df) / df['sender'].nunique(), 1))
```

```python
g = df.groupby('sender')

# previous message values, per vehicle
prev_posx = g['posx'].shift(1)
prev_posy = g['posy'].shift(1)
prev_time = g['sendTime'].shift(1)

# --- time gap between consecutive messages ---
df['dt'] = (df['sendTime'] - prev_time).astype('float32')

# --- distance jumped since last message ---
dx = df['posx'] - prev_posx
dy = df['posy'] - prev_posy
df['dist'] = np.sqrt(dx**2 + dy**2).astype('float32')

# --- claimed speed (magnitude of reported velocity) ---
df['claimed_spd'] = np.sqrt(df['spdx']**2 + df['spdy']**2).astype('float32')

# --- implied speed (distance / time) ---
df['implied_spd'] = (df['dist'] / df['dt'].replace(0, np.nan)).astype('float32')

# --- THE KEY FEATURE: do claimed and implied speed agree? ---
df['spd_mismatch'] = (df['claimed_spd'] - df['implied_spd']).abs().astype('float32')

# --- acceleration magnitude ---
df['acl_mag'] = np.sqrt(df['aclx']**2 + df['acly']**2).astype('float32')

# --- heading consistency: does heading match actual movement? ---
move_norm = np.sqrt(dx**2 + dy**2).replace(0, np.nan)
df['head_mismatch'] = (
    1 - ((dx / move_norm) * df['hedx'] + (dy / move_norm) * df['hedy'])
).astype('float32')

print("Features built.")
print("Memory:", round(df.memory_usage(deep=True).sum() / 1024**2, 1), "MB")
print()
print(df[['dt', 'dist', 'claimed_spd', 'implied_spd',
          'spd_mismatch', 'acl_mag', 'head_mismatch']].describe().round(3))
```

```python
feat_cols = ['dt', 'dist', 'claimed_spd', 'implied_spd',
             'spd_mismatch', 'acl_mag', 'head_mismatch']

# First message per vehicle has no predecessor -> NaN. Drop those rows.
before = len(df)
df = df.dropna(subset=feat_cols).reset_index(drop=True)
print(f"Dropped {before - len(df):,} first-message rows ({(before-len(df))/before*100:.2f}%)")

# Replace any remaining infinities (division edge cases)
df[feat_cols] = df[feat_cols].replace([np.inf, -np.inf], np.nan)
df = df.dropna(subset=feat_cols).reset_index(drop=True)

# Binary target: 0 = normal, 1 = any attack
df['is_attack'] = (df['class'] != 0).astype('int8')

print("Rows remaining:", len(df))
print()
print("Normal vs attack:")
print(df['is_attack'].value_counts())
print()
print("Feature means by class:")
print(df.groupby('is_attack')[feat_cols].mean().round(3).T)
```

```python
import time
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

X = df[feat_cols].values.astype('float32')
y = df['is_attack'].values

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y)

print("Train:", X_train.shape, " Test:", X_test.shape)

t0 = time.perf_counter()
rf = RandomForestClassifier(n_estimators=100, n_jobs=-1, random_state=42)
rf.fit(X_train, y_train)
print(f"Training took {time.perf_counter() - t0:.1f}s")

pred = rf.predict(X_test)
print()
print(f"Accuracy  {accuracy_score(y_test, pred):.4f}")
print(f"Precision {precision_score(y_test, pred):.4f}")
print(f"Recall    {recall_score(y_test, pred):.4f}")
print(f"F1        {f1_score(y_test, pred):.4f}")

print()
print("Feature importance:")
for n, i in sorted(zip(feat_cols, rf.feature_importances_),
                   key=lambda x: -x[1]):
    print(f"  {n:<16} {i:.4f}")
```

```python
import time
import numpy as np

# Warm up (first call includes one-time overhead)
_ = rf.predict(X_test[:100])

# Time 2000 individual single-message predictions
times = []
for i in range(2000):
    x = X_test[i:i+1]
    t0 = time.perf_counter()
    rf.predict(x)
    times.append((time.perf_counter() - t0) * 1000)   # ms

times = np.array(times)

print("PER-MESSAGE INFERENCE LATENCY (ms)")
print(f"  mean    {times.mean():.3f}")
print(f"  median  {np.percentile(times, 50):.3f}")
print(f"  p95     {np.percentile(times, 95):.3f}")
print(f"  p99     {np.percentile(times, 99):.3f}")
print(f"  max     {times.max():.3f}")

print()
print("3GPP TS 22.186 COMPLIANCE (judged on p99)")
p99 = np.percentile(times, 99)
for name, budget in [("Basic safety messaging", 100),
                     ("Platooning", 10),
                     ("Advanced / eV2X", 3)]:
    verdict = "PASS" if p99 <= budget else "FAIL"
    print(f"  {name:<24} {budget:>3} ms   {verdict}")
```

```python
import time, numpy as np

X1 = X_test[:1]

def bench(fn, n=2000, warmup=50):
    for _ in range(warmup): fn()
    t = []
    for _ in range(n):
        t0 = time.perf_counter(); fn(); t.append((time.perf_counter()-t0)*1000)
    return np.array(t)

# 1. baseline: what you just measured
a = bench(lambda: rf.predict(X1))

# 2. single-threaded (thread dispatch may cost more than it saves here)
rf.n_jobs = 1
b = bench(lambda: rf.predict(X1))

# 3. skip sklearn's input validation wrapper
import sklearn
with sklearn.config_context(assume_finite=True):
    c = bench(lambda: rf.predict(X1))

for name, arr in [("predict, n_jobs=-1", a),
                  ("predict, n_jobs=1 ", b),
                  ("assume_finite     ", c)]:
    print(f"{name}  median {np.percentile(arr,50):7.3f} ms   p99 {np.percentile(arr,99):7.3f} ms")

# 4. batch throughput — amortised per-message cost
t0 = time.perf_counter()
rf.predict(X_test[:100000])
batch_ms = (time.perf_counter()-t0)*1000/100000
print(f"\nbatched (100k)      per-message {batch_ms:.4f} ms")
```

```python
import time, numpy as np, json
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score
import sklearn

tree_counts = [10, 25, 50, 100, 200]
SEEDS = [0, 1, 2]
sweep = []

for n in tree_counts:
    accs, f1s, meds, p99s = [], [], [], []

    for seed in SEEDS:
        m = RandomForestClassifier(n_estimators=n, n_jobs=-1, random_state=seed)
        m.fit(X_train, y_train)

        pred = m.predict(X_test)
        accs.append(accuracy_score(y_test, pred))
        f1s.append(f1_score(y_test, pred))

        m.n_jobs = 1
        X1 = X_test[:1]
        with sklearn.config_context(assume_finite=True):
            for _ in range(100):
                m.predict(X1)
            t = []
            for _ in range(2000):
                s = time.perf_counter(); m.predict(X1)
                t.append((time.perf_counter() - s) * 1000)
        t = np.array(t)
        meds.append(np.percentile(t, 50))
        p99s.append(np.percentile(t, 99))

    row = {
        "trees": n,
        "acc_mean": round(float(np.mean(accs)), 5),
        "acc_std":  round(float(np.std(accs)), 5),
        "f1_mean":  round(float(np.mean(f1s)), 5),
        "median_ms_mean": round(float(np.mean(meds)), 3),
        "p99_ms_mean":    round(float(np.mean(p99s)), 3),
        "p99_ms_std":     round(float(np.std(p99s)), 3),
    }
    sweep.append(row)
    print(f"{n:>4} trees | acc {row['acc_mean']:.5f} ±{row['acc_std']:.5f} | "
          f"median {row['median_ms_mean']:6.3f} | p99 {row['p99_ms_mean']:6.3f} ±{row['p99_ms_std']:.3f}")

with open(r"C:\projects\v2x-detection\outputs\sweep_v2.json", "w") as f:
    json.dump(sweep, f, indent=2)

print("\nSaved sweep_v2.json")
```

```python

import time
import numpy as np
from sklearn.metrics import (accuracy_score, precision_score,
                             recall_score, f1_score)
import sklearn

# Physical plausibility thresholds (urban vehicle limits)
MAX_SPD_MISMATCH = 5.0     # m/s — claimed vs implied speed may differ by this
MAX_SPEED        = 60.0    # m/s ~ 216 km/h
MAX_ACCEL        = 10.0    # m/s^2 ~ 1g
MIN_DT           = 0.5     # s — faster than this implies flooding

idx = {c: i for i, c in enumerate(feat_cols)}

def rule_predict(X):
    """Flag a message if it violates any physical plausibility check."""
    flag = (
        (X[:, idx['spd_mismatch']] > MAX_SPD_MISMATCH) |
        (X[:, idx['implied_spd']]  > MAX_SPEED) |
        (X[:, idx['acl_mag']]      > MAX_ACCEL) |
        (X[:, idx['dt']]           < MIN_DT)
    )
    return flag.astype(np.int8)

pred_rule = rule_predict(X_test)

print("RULE-BASED BASELINE")
print(f"  Accuracy  {accuracy_score(y_test, pred_rule):.4f}")
print(f"  Precision {precision_score(y_test, pred_rule):.4f}")
print(f"  Recall    {recall_score(y_test, pred_rule):.4f}")
print(f"  F1        {f1_score(y_test, pred_rule):.4f}")

# latency, same methodology as the model
X1 = X_test[:1]
for _ in range(100):
    rule_predict(X1)
t = []
for _ in range(2000):
    s = time.perf_counter(); rule_predict(X1)
    t.append((time.perf_counter() - s) * 1000)
t = np.array(t)
print(f"  median    {np.percentile(t,50):.4f} ms")
print(f"  p99       {np.percentile(t,99):.4f} ms")
```

```python
import time, numpy as np, joblib, os, json
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score
import sklearn

tmp = r"C:\projects\v2x-detection\outputs\_tmp.joblib"
results = []

for depth in [6, 8, 10, 12, 15, None]:
    m = RandomForestClassifier(n_estimators=10, max_depth=depth,
                               n_jobs=-1, random_state=0)
    m.fit(X_train, y_train)
    pred = m.predict(X_test)

    joblib.dump(m, tmp, compress=3)
    size_mb = os.path.getsize(tmp) / 1024**2

    m.n_jobs = 1
    X1 = X_test[:1]
    with sklearn.config_context(assume_finite=True):
        for _ in range(100):
            m.predict(X1)
        t = []
        for _ in range(1500):
            s = time.perf_counter(); m.predict(X1)
            t.append((time.perf_counter() - s) * 1000)
    t = np.array(t)

    r = {"max_depth": depth,
         "accuracy": round(accuracy_score(y_test, pred), 5),
         "f1": round(f1_score(y_test, pred), 5),
         "size_mb": round(size_mb, 2),
         "p99_ms": round(float(np.percentile(t, 99)), 3)}
    results.append(r)
    print(f"depth {str(depth):>4} | acc {r['accuracy']:.5f} | f1 {r['f1']:.5f} | "
          f"size {r['size_mb']:7.2f} MB | p99 {r['p99_ms']:6.3f} ms")

os.remove(tmp)
with open(r"C:\projects\v2x-detection\outputs\depth_sweep.json", "w") as f:
    json.dump(results, f, indent=2)
print("\nSaved depth_sweep.json")
```
